Visualization 2: Washington Post Active Shooters

Inconclusion, data visualizations cannot be completely neutral because creators must make choices about which data to include and how to represent it. A visualization can still be factual and honest without being neutral. The important issue is whether the design choices accurately represent the data and are appropriate for the intended purpose.

When choosing a visualization, I should consider:

- Intended purpose: persuading, comparing, evaluating, or exploring.
- Intended audience: age, education, expertise, and accessibility needs.
- Intended medium: print, web, poster, or presentation.
- Data and attribute types.
- Cognitive load and the audience's familiarity with the chart type.

Python Graph Gallery Activity

Visualization name: Basic Connected Scatterplot with Matplotlib

Source: https://python-graph-gallery.com/132-basic-connected-scatterplot/

Libraries used: Matplotlib, NumPy, and pandas

A connected scatterplot combines lines and markers. This example presents two numerical series on the same axes and uses different marker and line styles to distinguish them.

```python
import matplotlib.pyplot as plt 
import numpy as np 
import pandas as pd
```

```python
# Set the default figure size. 
plt.rcParams["figure.figsize"] = (10, 6) 
# Create a random-number generator with a fixed seed for reproducibility. 
rng = np.random.default_rng(1111) 
# Generate sample data. 
x = np.array(range(10)) 
y = rng.integers(10, 100, 10) 
z = y + rng.integers(5, 20, 10)
```

Reproducing the Connected Scatterplot

The two series are drawn on the same axes. Different line styles, markers, marker sizes, and marker fills are used to distinguish the series.

```python
plt.plot( x, z, linestyle="--", linewidth=3, marker="X", markersize=10, markerfacecolor="red", markeredgecolor="black", label="Income", ) 
plt.plot( x, y, linestyle=":", marker="o", markersize=15, markerfacecolor="None", label="Expenses", ) 
plt.legend()
```

Evaluation of the Selected Visualization
Aesthetic Quality

The visualization is visually engaging because the two series use distinct marker shapes, sizes, and line styles. The contrast between the filled X markers and hollow circular markers makes the series easy to distinguish. However, the large markers make the graph look somewhat busy, and a more restrained marker size could produce a cleaner appearance.

Substantive Quality

The legend clearly identifies the two series, and the fixed random seed makes the example reproducible. However, the graph uses randomly generated sample data rather than an observed dataset. It also lacks a descriptive title, axis labels, units, and an explanation of what the x-axis represents. These missing elements limit the conclusions that a viewer can draw from the graph.

Perceptual Quality

The connected lines make it easy to follow changes across the ordered x-values, while the different marker shapes and line styles allow the series to be distinguished without relying only on colour. The legend is clear and does not cover the main data. However, the overall message is not immediately clear because the chart does not provide a title or meaningful axis labels.
