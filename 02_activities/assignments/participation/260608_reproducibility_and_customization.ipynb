Reproducibility Notes

Reproducible work makes the data, code, and analytical methods available so that another person can repeat the process and generate the same result.

Reproducibility supports transparency and accountability, but a reproducible result is not automatically a correct result.

Programmatic work, plain-text code, clear documentation, and useful comments make visualizations easier to reproduce, maintain, and update.

```python
import numpy as np 
import matplotlib.pyplot as plt 
import pandas as pd 
import scipy 
import PIL 
import requests 
```

```python
np.random.seed(613) 
x = np.arange(50) 
y1 = np.random.randint(0, 100, 50) 
y2 = np.random.randint(0, 100, 50)
```

Both y variables are plotted on the same axes so they can be compared directly.

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
# Plot both series on the same axes. 
ax.plot(x, y1) 
ax.plot(x, y2)
```

Adding a Legend

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x, y1, label="Person 1") 
ax.plot(x, y2, label="Person 2") 
# Place the legend in the lower-right corner. 
ax.legend(loc="lower right")
```

Customizing the Legend

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x, y1, label="Person 1") 
ax.plot(x, y2, label="Person 2") 
ax.legend( loc="lower right", frameon=True, fontsize=12, ncol=2, shadow=True, )
fig.show()
```

Moving the Legend Outside the Plot

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.plot(x, y1, label="Person 1") 
ax.plot(x, y2, label="Person 2") 
ax.legend( loc="upper left", bbox_to_anchor=(1, 1), ) 
fig.show()
```

Annotation

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
# Place text at the data coordinate x=10, y=95. 
ax.text( 10, 95, "This value is important!", ) 
fig.show()
```

Customizing Annotations

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
ax.text( 10, 95, "This value is important!", ha="center", color="red", size=20, ) 
fig.show()
```

Positioning Text

```python
fig, ax = plt.subplots() 
ax.axis([0, 10, 0, 10]) 

ax.text( 1, 5, "Data: (1, 5)", transform=ax.transData, )
ax.text( 0.5, 0.1, "Axes: (0.5, 0.1)", transform=ax.transAxes, )
ax.text( 0.2, 0.2, "Figure: (0.2, 0.2)", transform=fig.transFigure, ) 
fig.show()
```

Annotating with Arrows

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
ax.annotate( "This is important!", xy=(10, 95), xytext=(20, 94), arrowprops={ "facecolor": "black", }, ) 
fig.show()
```

Customizing Arrows

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
ax.annotate( "This is important!", xy=(10, 95), xytext=(20, 94), arrowprops={ "arrowstyle": "wedge", "color": "hotpink", }, ) 
fig.show()
```

Removing Tick Marks and Labels

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
# Remove y-axis major ticks and their labels. 
ax.yaxis.set_major_locator(NullLocator()) 
# Remove x-axis labels while leaving the tick positions. 
ax.xaxis.set_major_formatter(NullFormatter()) 
fig.show()
```

Limiting Tick Marks

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
# Display no more than approximately three major x-axis ticks. 
ax.xaxis.set_major_locator(MaxNLocator(3)) 
fig.show()
```

Setting Tick Intervals

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
ax.xaxis.set_major_locator(MultipleLocator(5)) 
fig.show()
```

Rotating Tick Labels

```python
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
plt.xticks( rotation=45, ha="right", ) 
```

Activity: Modifying the X-Axis Title Font

For this activity, the x-axis title uses a serif font, an indigo colour, and a font size of 18.

```python
font1 = { "family": "serif", "color": "indigo", } 
fig, ax = plt.subplots(figsize=(5, 3)) 
ax.scatter(x, y1, label="Person 1") 
ax.scatter(x, y2, label="Person 2") 
ax.legend(loc="lower right") 
plt.xlabel( "Shiny New X Axis!", fontsize=18, fontdict=font1, ) 
fig.show()
```
